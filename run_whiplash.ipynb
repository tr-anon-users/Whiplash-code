# Whiplash · AISTATS 실행 예시

논문 식 (1)–(10)에 맞춘 현재 API를 사용합니다. 먼저 README의 가상환경과 의존성을 설치하고 해당 환경을 notebook kernel로 선택하세요. 원본 데이터 경로는 아래 설정에서 변경할 수 있습니다.

이 notebook은 실험 계획 확인과 2개 링크의 제한된 smoke 실행용입니다. 저장된 출력이나 사전 계산한 성능 결과는 포함하지 않습니다.


```python
from pathlib import Path
import subprocess
import sys

project_dir = Path.cwd()
if not (project_dir / "train.py").is_file():
    project_dir = Path(r"G:\내 드라이브\03 연구\03 Tramba\Github\_AISTATS")
speed_file = Path(r"G:\내 드라이브\03 연구\03 Tramba\data\202406_01_05_FF.csv")
link_file = Path(r"G:\내 드라이브\03 연구\03 Tramba\data\강남구 링크_sample.csv")

for path in (project_dir / "train.py", speed_file, link_file):
    if not path.is_file():
        raise FileNotFoundError(f"실제 파일 경로를 확인하세요: {path}")
print("Python:", sys.executable)
print("Project:", project_dir)
```

```python
# 계획만 출력합니다. --dry-run은 run_experiments.py 전용입니다.
plan_command = [
    sys.executable, "run_experiments.py",
    "--data-path", str(speed_file),
    "--link-path", str(link_file),
    "--speed-unit", "kmh",
    "--stage", "all",
    "--horizons", "12",
    "--dry-run",
]
subprocess.run(plan_command, cwd=project_dir, check=True)
```

## 제한된 학습 확인

다음 셀은 현재 Python 환경으로 1 epoch, 분할별 최대 2 batch를 실행하고 `outputs/notebook_smoke` 아래에 결과를 저장합니다. 이 결과는 논문 성능 재현이 아닌 실행 점검입니다. 전체 실험에는 README의 `run_experiments.py` 명령을 사용하세요. 기본 프로토콜은 앞쪽 개발 구간의 5-fold 튜닝 후 시간순 70/10/20 분할에서 horizon별 10개 seed를 평가합니다.


```python
smoke_command = [
    sys.executable, "train.py",
    "--data-path", str(speed_file),
    "--link-path", str(link_file),
    "--speed-unit", "kmh",
    "--pred-len", "12",
    "--epochs", "1",
    "--batch-size", "4",
    "--max-train-batches", "2",
    "--max-val-batches", "2",
    "--max-test-batches", "2",
    "--device", "cpu",
    "--output-dir", "outputs/notebook_smoke",
]
subprocess.run(smoke_command, cwd=project_dir, check=True)
```
